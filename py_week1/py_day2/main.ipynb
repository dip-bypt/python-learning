```python
# Import the pandas library, which is used for data manipulation and analysis
import pandas as pd

# Read the CSV file named 'students.csv' into a DataFrame called df
# The CSV file should have columns like 'name' and 'age'
df = pd.read_csv('students.csv')

# Print the entire DataFrame to see the contents of the CSV file
print(df)
```

```python
# Use a list comprehension to create a list of names of students older than 18
# zip(df['name'], df['age']) pairs each name with the corresponding age
# The condition 'if age > 18' filters only those students whose age is greater than 18
names_over_18 = [name for name, age in zip(df['name'], df['age']) if age > 18]

# Print the list of names of students older than 18
print(names_over_18)
```
